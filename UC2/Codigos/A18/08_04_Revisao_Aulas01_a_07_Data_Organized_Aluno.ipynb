```python
# Tabela auxiliar simulando extração do banco MySQL (dim_regioes)
df_gerentes = pd.DataFrame({
    'Regiao_Cliente': ['Sul', 'Sudeste', 'Centro-Oeste', 'Nordeste', 'Norte'],
    'Gerente_Regional': ['Carlos Mendes', 'Fernanda Lima', 'Roberto Alves', 'Juliana Costa', 'Marcos Souza'],
    'Meta_Ticket_BRL': [4800.00, 5200.00, 4500.00, 4700.00, 4600.00]
})

# Desenvolva aqui o agrupamento por região e o cruzamento via pd.merge():
# SQL Query
print("O SQL Query: SELECT Regiao_Cliente, SUM(Valor_Contrato_BRL) FROM {nome_da_tabela} GROUP BY Regiao_Cliente")

# merge
df_res_reg = df.groupby(by='Regiao_Cliente')['Valor_Contrato_BRL'].mean().round(2)
df_merge = pd.merge(
    df_res_reg,
    df_gerentes,
    how= 'left',
    on='Regiao_Cliente'
)

df_merge["Superaram_Media"] = df_merge["Valor_Contrato_BRL"] > df_merge["Meta_Ticket_BRL"]
display(df_merge)
```

## Etapa 4 (Revisão Aula 05): NumPy Arrays e Confiabilidade da Média

### Perguntas Norteadoras
1. Converta a coluna `Valor_Contrato_BRL` em um **Array NumPy** e calcule a média (`np.mean`) e a mediana (`np.median`).
2. Calcule a distância relativa entre média e mediana pela fórmula do PTD: $\text{Distância} = \frac{\text{Média} - \text{Mediana}}{\text{Mediana}}$.
3. Sabendo que uma distância $\ge |25\%|$ indica **alta influência de valores extremos**, a média é uma medida confiável para o diretor apresentar ao conselho? A distorção é puxada por valores maiores ou menores?

```python
import numpy as np

# Desenvolva aqui o código da Etapa 4 (np.array, mean, median e regra de distância):
# coluna `Valor_Contrato_BRL` em um **Array NumPy** e calcule a média (`np.mean`) e a mediana (`np.median`)
np_array = np.array(df["Valor_Contrato_BRL"])
print(np_array)

print(f"A média é {np_array.mean():.2f}")
print(f"A mediana é {np.median(np_array):.2f}")
print("\n" + (80*"=") + "\n")

# Calcule a distância relativa entre média e mediana pela fórmula do PTD
dis_rel = (np.mean(np_array) - np.median(np_array))/np.median(np_array)
print(f"A distância relativa entre média e mediana pela fórmula do PTD é {dis_rel}")
print("\n" + (80*"=") + "\n")

print("A distancia relativa é positiva, então a media é maior do que a mediana, então a assimetria é positiva")
```
