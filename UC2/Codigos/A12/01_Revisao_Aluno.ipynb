```python
import pandas as pd # biblioteca para manipulação de dados 
import numpy as np # biblioteca para operações numéricas

# Definindo número aleatório para reprodutibilidade dos dados
np.random.seed(42)

# Definindo o volume de linhas do arquivo
TOTAL_LINHAS = 1000

# Lista de categorias e produtos para sorteio
categorias = ['Eletrônicos', 'Móveis', 'Informática', 'Eletrodomésticos']
cidades = ['Rio de Janeiro', 'Niterói', 'Petrópolis', 'Volta Redonda', 'Macaé']

# Gerando dados sintéticos
dados_sinteticos = {
    'ID_Transacao': [f"TXN-{10000 + i}" for i in range(TOTAL_LINHAS)], # 1000 + i: lista incrementando por 1 acada item ate o final
    'Data_Venda': pd.date_range(start='2024-01-01', periods=TOTAL_LINHAS, freq='4h'), #lista de datas comecando no 1/1/24 e incrementando por 4h
    'Cidade': np.random.choice(cidades, size=TOTAL_LINHAS, p=[0.4, 0.2, 0.15, 0.15, 0.1]),
    'Categoria': np.random.choice(categorias, size=TOTAL_LINHAS),
    'Quantidade': np.random.randint(1, 10, size=TOTAL_LINHAS),
    'Preco_Unitario': np.round(np.random.uniform(50.0, 3500.0, size=TOTAL_LINHAS), 2)
}

# Criando o DataFrame
df_gerado = pd.DataFrame(dados_sinteticos)

# Calculando a coluna do Valor Total
df_gerado['Valor_Total'] = df_gerado['Quantidade'] * df_gerado['Preco_Unitario']

```

Exercício 1: Análise de Vendas
Enunciado:
Crie um programa que analise um conjunto de dados de vendas. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_vendas(nome_arquivo): Esta função deve ler um arquivo CSV contendo informações de vendas com as colunas: Produto, Quantidade, Preço e retornar um DataFrame.

Função calcula_receita(df): Esta função deve receber um DataFrame e calcular a receita total de cada produto (multiplicando Quantidade por Preço). A receita deve ser adicionada como uma nova coluna chamada Receita.

Função total_vendas_por_produto(df): Esta função deve agrupar os dados por produto e calcular o total de vendas (soma das quantidades) e a receita total (soma das receitas). Retorne um novo DataFrame contendo o total de vendas e a receita por produto.

Função filtra_produtos_altas_vendas(df, limite): Esta função deve filtrar o DataFrame para retornar apenas os produtos cuja receita total é superior a um valor limite passado como argumento.

Função analise_vendas(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores, retornando o DataFrame final com os produtos que geraram mais de R$ 500,00 em receita.

```python
import pandas as pd
from pathlib import Path


## le_dados_vendas(nome_arquivo): Esta função deve ler um arquivo CSV contendo informações de vendas com as 
#  colunas: Produto, Quantidade, Preço e retornar um DataFrame.

def le_dados_vendas(file_name):
    uc_2 = Path.cwd().parent.parent
    path = (fr"{uc_2}\DATA\{file_name}")
    return pd.read_csv(path, sep=",", encoding="utf-8")

## calcula_receita(df): Esta função deve receber um DataFrame e calcular a receita total de 
#  cada produto (multiplicando Quantidade por Preço). A receita deve ser adicionada como uma nova coluna chamada Receita.

def calcula_receita(df):
    df["Receita"] = df["Quantidade"] * df["Preco"]
    pass

## total_vendas_por_produto(df): Esta função deve agrupar os dados por produto e
#  calcular o total de vendas (soma das quantidades) e a receita total (soma das receitas). 
#  Retorne um novo DataFrame contendo o total de vendas e a receita por produto.

def total_vendas_por_produto(df):
    #pegando os produtos unicos 
    lst_produtos = sorted(list(set(df["Produto"])))
    
    lst_sum_quantidades = []
    lst_sum_receitas = []

    #looping pelos produtos unicos para somar suas quantidades e receitas
    for i in lst_produtos:
        temp = df[df["Produto"] == i]
        lst_sum_quantidades.append(temp["Quantidade"].sum())
        lst_sum_receitas.append(temp["Receita"].sum())

    return pd.DataFrame({"Produto": lst_produtos, "Total de Vendas": lst_sum_quantidades, "Total de Recitas": lst_sum_receitas})

## filtra_produtos_altas_vendas(df, limite): Esta função deve filtrar o DataFrame para retornar apenas os 
#  produtos cuja receita total é superior a um valor limite passado como argumento.

def filtra_produtos_altas_vendas(df, limite, receita="Receita"):
    return df[df[receita] > limite]

## analise_vendas(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores, retornando o 
#  DataFrame final com os produtos que geraram mais de R$ 500,00 em receita.

def analise_vendas(path, limite=500, receita="Receita"):
    df_vendas = le_dados_vendas(path)
    calcula_receita(df_vendas)
    #print(df_vendas)
    df_total_vendas = total_vendas_por_produto(df_vendas)
    #print(df_total_vendas)
    df_filtro = filtra_produtos_altas_vendas(df_total_vendas, limite, receita=receita)
    #print(df_filtro)
    return df_filtro

print(analise_vendas("vendas.csv", receita="Total de Recitas"))
```

```python
import pandas as pd

# le_dados_vendas(nome_arquivo): Esta função deve ler um arquivo CSV contendo informações de vendas com as 
# colunas: Produto, Quantidade, Preço e retornar um DataFrame.
def le_dados_vendas(caminho):
    df = pd.read_csv(caminho)
    return df

# calcula_receita(df): Esta função deve receber um DataFrame e calcular a receita total de 
# cada produto (multiplicando Quantidade por Preço). A receita deve ser adicionada como uma nova coluna chamada Receita.
def calcula_receita(df):
    df["Receita"] = df["Preco"] * df["Quantidade"]
    pass

# total_vendas_por_produto(df): Esta função deve agrupar os dados por produto e calcular o total de vendas (soma das quantidades) 
# e a receita total (soma das receitas). Retorne um novo DataFrame contendo o total de vendas e a receita por produto.
def total_vendas_por_produto(df):
    #lista de produtos unicos
    produtos = sorted(list(set(df["Produto"])))

    lst_quantidades = []
    lst_receitas = []

    for produto in produtos:
        temp = df[df["Produto"]  == produto]
        lst_quantidades.append(temp["Quantidade"].sum())
        lst_receitas.append(temp["Receita"].sum())

    df_novo = pd.DataFrame({
        "Produto": produtos,
        "Quantidade Total": lst_quantidades,
        "Receita Total": lst_receitas
    })
    
    return df_novo

# filtra_produtos_altas_vendas(df, limite): Esta função deve filtrar o DataFrame para retornar apenas os produtos 
# cuja receita total é superior a um valor limite passado como argumento.
def filtra_produtos_altas_vendas(df, limite):
    df_filtro = df[df["Receita Total"] > limite]
    return df_filtro

# analise_vendas(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores, 
# retornando o DataFrame final com os produtos que geraram mais de R$ 500,00 em receita.
def analise_vendas(caminho):
    df = le_dados_vendas(caminho)
    calcula_receita(df)

    df_novo = total_vendas_por_produto(df)

    df_fil = filtra_produtos_altas_vendas(df_novo, 50)

    return df_fil


```

Função calcula_media(df): Esta função deve calcular a média das notas de cada aluno e adicionar uma nova coluna chamada Media ao DataFrame.

Função verifica_aprovacao(df): Esta função deve adicionar uma nova coluna chamada Aprovado, que deve conter True se a média do aluno for maior ou igual a 7 e False caso contrário.

Função separa_alunos_aprovados_reprovados(df): Esta função deve separar os alunos em dois DataFrames: um para aprovados e outro para reprovados, com base na coluna Aprovado.

Função analise_notas(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores e retornar os DataFrames de alunos aprovados e reprovados.

Exercício 3: Análise de Dados de Funcionários
Enunciado:
Crie um programa que analise os dados de funcionários de uma empresa. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_funcionarios(nome_arquivo): Esta função deve ler um arquivo CSV com as colunas: Nome, Departamento, Salário e retornar um DataFrame.

Função calcula_salario_medio(df): Esta função deve agrupar os dados por departamento e calcular o salário médio de cada um, retornando um novo DataFrame com o nome do departamento e o salário médio.

Função filtra_departamentos_alto_salario(df, limite): Esta função deve filtrar o DataFrame para retornar apenas os departamentos cuja média salarial é superior a um valor limite passado como argumento.

Função analise_funcionarios(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores e retornar o DataFrame final com os departamentos que têm um salário médio acima de R$ 5000,00.

```python
import pandas as pd
from pathlib import Path

# le_dados_funcionarios(nome_arquivo): Esta função deve ler um arquivo CSV com as 
# colunas: Nome, Departamento, Salário e retornar um DataFrame.
def le_dados_funcionarios(file_name, sep=",", enc="utf-8"):
    uc_2 = Path.cwd().parent.parent
    path = (fr"{uc_2}\DATA\{file_name}")
    return pd.read_csv(path, sep=sep, encoding=enc)


# calcula_salario_medio(df): Esta função deve agrupar os dados por departamento e 
# calcular o salário médio de cada um, retornando um novo DataFrame com o nome do departamento e o salário médio.
def calcula_salario_medio(df):
    lst_dep = sorted(list(set(df["Departamento"])))
    lst_sal_med = []

    for i in lst_dep:
        temp = df[df["Departamento"] == i]
        lst_sal_med.append(temp["Salário"].sum()/len(temp))

    return pd.DataFrame({"Departamento": lst_dep, "Salário Medio": lst_sal_med})


# filtra_departamentos_alto_salario(df, limite): Esta função deve filtrar o DataFrame para
#  retornar apenas os departamentos cuja média salarial é superior a um valor limite passado como argumento.
def filtra_departamentos_alto_salario(df, limite):
    return df[df["Salário Medio"] > limite]


# analise_funcionarios(nome_arquivo): Esta função deve orquestrar as chamadas das 
# funções anteriores e retornar o DataFrame final com os departamentos que têm um salário médio acima de R$ 5000,00.
def analise_funcionarios(path, limite):
    df_fun = le_dados_funcionarios(path)
    df_dep = calcula_salario_medio(df_fun)
    
    return filtra_departamentos_alto_salario(df_dep, limite)

```

```python
import pandas as pd

# le_dados_funcionarios(nome_arquivo): Esta função deve ler um arquivo CSV com as 
# colunas: Nome, Departamento, Salário e retornar um DataFrame.
def le_dados_funcionarios(caminho):
    df = pd.read_csv(caminho, sep=",", encoding="utf-8")
    return df

# calcula_salario_medio(df): Esta função deve agrupar os dados por departamento e 
# calcular o salário médio de cada um, retornando um novo DataFrame com o nome do departamento e o salário médio.
def calcula_salario_medio(df):
    deps = sorted(list(set(df["Departamento"])))

    lst_sal = []

    for dep in deps:
        temp = df[df["Departamento"]  == dep]
        lst_sal.append(temp["Salário"].sum()/len(temp))

    df_novo = pd.DataFrame({
        "Departamento": deps,
        "Salário Médio": lst_sal
    })
    
    return df_novo
    

# filtra_departamentos_alto_salario(df, limite): Esta função deve filtrar o DataFrame para 
# retornar apenas os departamentos cuja média salarial é superior a um valor limite passado como argumento.
def filtra_departamentos_alto_salario(df, limite):
    return df[df["Salário Médio"] > limite]

# analise_funcionarios(nome_arquivo): Esta função deve orquestrar as chamadas das funções 
# anteriores e retornar o DataFrame final com os departamentos que têm um salário médio acima de R$ 5000,00.
def analise_funcionarios(caminho):
    df = le_dados_funcionarios(caminho)
    df_novo = calcula_salario_medio(df)
    df_filtro = filtra_departamentos_alto_salario(df_novo, 5000)
    return df_filtro
```

```python
import pandas as pd
from pathlib import Path

# ler o arquivo 
def ler(file_name, sep=",", enc="utf-8"):
    uc_2 = Path.cwd().parent.parent
    path = (fr"{uc_2}\DATA\{file_name}")
    return pd.read_csv(path, sep=sep,encoding=enc)


# calcular as receitas
def calc_receitas(df):
    df["Receita"] = df["Quantidade"] * df["Preço"]
    pass


#  resumir e filtrar os resultados.
def res_filtro(df, filtro=0):
    #pegando os produtos unicos 
    lst_produtos = sorted(list(set(df["Produto"])))

    lst_sum_quantidades = []
    lst_sum_receitas = []

    #looping pelos produtos unicos para somar suas quantidades e receitas
    for i in lst_produtos:
        temp = df[df["Produto"] == i]
        lst_sum_quantidades.append(temp["Quantidade"].sum())
        lst_sum_receitas.append(temp["Receita"].sum())

    df_resumo = pd.DataFrame({"Produto": lst_produtos, "Total de Vendas": lst_sum_quantidades, "Total de Recitas": lst_sum_receitas})
    if not filtro:
        return df_resumo

    return df_resumo[df_resumo["Total de Recitas"] > filtro]

#main
def main():
    df = ler("vendas_2.csv")
    print('\n', ('='*80), '\n')
    print(df)
    print('\n', ('='*80), '\n')
    calc_receitas(df)
    print(df)
    print('\n', ('='*80), '\n')
    df_resumo = res_filtro(df, 1000)
    print(df_resumo)
    print('\n', ('='*80), '\n') 



if __name__ == "__main__":
    main()
```
